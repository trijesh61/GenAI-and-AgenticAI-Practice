```python
import chromadb

client = chromadb.Client()
```

```python
client
```

```python
collection = client.create_collection(name="news11")
```

```python
collection.add(
documents=[
"Apple reported its quarterly earnings today.",
"Apple has a lot of vitamin A"
],
ids=["id1", "id2"]
)
```

```python
result=collection.query(
    query_texts=["I phone 16 will launching soon"],
    n_results=2
    )
result
```

```python
import chromadb

clientp = chromadb.PersistentClient(path="db")
```

```python
from chromadb.utils import embedding_functions
ef= embedding_functions.SentenceTransformerEmbeddingFunction(model_name="sentence-transformers/all-MiniLM-L12-v2")
collection = clientp.create_collection(name="news_v201",embedding_function=ef)
```

```python
collection.add(
documents=[
"Apple reported its quarterly earnings today.",
"Apple has a lot of vitamin A"
],
ids=["id1", "id2"]
)
```

```python
collection.add(
documents=[
"Apple reported its quarterly earnings today.",
"Apple has a lot of vitamin A"
],
ids=["id1", "id2"]
)
```

```python
result=collection.query(
    query_texts=["I phone 16 will launching soon"],
    n_results=2
    )
result
```
