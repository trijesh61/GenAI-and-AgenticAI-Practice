```python
import os
from dotenv import load_dotenv
from crewai import LLM

load_dotenv()

llm = LLM(
    model=os.getenv("TOKENHARBOR_MODEL"),
    api_key=os.getenv("TOKENHARBOR_API_KEY"),
    base_url=os.getenv("TOKENHARBOR_BASE_URL"),
    custom_openai=True,
    temperature=0.7,
)
```

```python
from crewai import Agent,Task,Crew

research_agent = Agent(
  role = "Research Speacalist",
  goal = "Research intresting facts about the topic : {topic}",
  backstory = "You are an expert in finding the relevant and factual data",
  verbose = True,
  llm = llm
)

Writer_agent = Agent(
  role ="Creative Writer",
  goal = "Wrirte a short blog summary using the research",
  backstory = "you are skilled at writing engaging summaries based on the provided content",
  verbose = True,
  llm  = llm
)
```

```python
task1 = Task(
                description="Find 3-5 intresting and recent facts about {topic}",
                expected_output= "A Bullet List of 3-5 facts",
                agent = research_agent
)

task2 = Task(
  description= "Write a 100-word Blog post summary about the {topic} using  the facts from the research",
  expected_output= " A blog post summary",
  agent = Writer_agent,
  context= [task1]
)


crew = Crew(
  agents = [research_agent, Writer_agent],
  tasks = [task1, task2],
  verbose= True,
)
```

```python
await crew.kickoff_async(inputs={'topic' : 'The Future of Data Science in this AI Era'})
```

```python
from crewai import Agent,Task,Crew
from crewai_tools import SerperDevTool

research_agent = Agent(
  role = "Research Speacalist",
  goal = "Research intresting facts about the topic : {topic}",
  backstory = "You are an expert in finding the relevant and factual data",
  tools = [SerperDevTool()],
  verbose = True,
  llm = llm
)

Writer_agent = Agent(
  role ="Creative Writer",
  goal = "Wrirte a short blog summary using the research",
  backstory = "you are skilled at writing engaging summaries based on the provided content",
  verbose = True,
  llm  = llm
)
```

```python
task1 = Task(
                description="Find 3-5 intresting and recent facts about {topic} as on october 2026",
                expected_output= "A Bullet List of 3-5 facts",
                agent = research_agent
)

task2 = Task(
  description= "Write a 100-word Blog post summary about the {topic} using  the facts from the research",
  expected_output= " A blog post summary",
  agent = Writer_agent,
  context= [task1]
)


crew = Crew(
  agents = [research_agent, Writer_agent],
  tasks = [task1, task2],
  verbose= True,
)
```
