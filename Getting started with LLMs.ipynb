```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.exceptions import OutputParserException
from dotenv import load_dotenv
load_dotenv()
```

```python
prompt='''' From the below news article ,  Extract the Revenue and eps in Json format containg the following keys
        : 'revenue_actual','revenue_expected',eps_actual', 'eps_expected' 
        only give the valid json no preamble

        Article
        =========
        {article}



'''
pt=PromptTemplate.from_template(prompt)
pt
```

```python
llm=ChatGroq(model="llama-3.1-8b-instant")
chain= pt | llm
chain
```

```python
Article=''' Here’s how Apple did versus LSEG consensus estimates for the quarter ending Sept. 27: 

EPS: $1.85 vs. $1.77 estimated  
Revenue: $102.47 billion vs. $102.24 billion estimated   '''
```

```python
res=chain.invoke({"article":"Article"}) 
```
